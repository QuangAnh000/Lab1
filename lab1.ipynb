# 4. Phân loại dữ liệu định tính và định lượng

```python
dinh_tinh = ['GT', 'DT', 'KV', 'KT']
dinh_luong = [f'{m}{n}' for n in [1,2,6] for m in ['T','L','H','S','V','X','D','N']] + ['DH1', 'DH2', 'DH3']
print("\n🔸 Dữ liệu định tính:", dinh_tinh)
print("🔸 Dữ liệu định lượng:", dinh_luong)
```

# 5. Xử lý thiếu cột DT

```python
print("\n📊 Thống kê DT:")
print(df['DT'].value_counts(dropna=False))
df['DT'].fillna(0, inplace=True)
```

# 6. Xử lý thiếu T1 bằng mean

```python
print("\n📊 Thiếu dữ liệu T1:", df['T1'].isna().sum())
df['T1'].fillna(df['T1'].mean(), inplace=True)
```

# 7. Xử lý thiếu tất cả điểm còn lại

```python
for col in dinh_luong:
    df[col].fillna(df[col].mean(), inplace=True)
```

# 8. Tạo TBM1, TBM2, TBM3

```python
def tinh_tb(n):
    return (df[f'T{n}']*2 + df[f'L{n}'] + df[f'H{n}'] + df[f'S{n}'] +
            df[f'V{n}']*2 + df[f'X{n}'] + df[f'D{n}'] + df[f'N{n}']) / 10

df['TBM1'] = tinh_tb(1)
df['TBM2'] = tinh_tb(2)
df['TBM3'] = tinh_tb(6)
```

# 9. Xếp loại học lực theo thang điểm

```python
def xeploai(tb):
    if tb < 5: return 'Y'
    elif tb < 6.5: return 'TB'
    elif tb < 8.0: return 'K'
    elif tb < 9.0: return 'G'
    else: return 'XS'

df['XL1'] = df['TBM1'].apply(xeploai)
df['XL2'] = df['TBM2'].apply(xeploai)
df['XL3'] = df['TBM3'].apply(xeploai)
```

# 10. Chuyển thang điểm 4 theo min-max normalization

```python
for col in ['TBM1', 'TBM2', 'TBM3']:
    new_col = 'US_' + col
    df[new_col] = (df[col] - df[col].min()) / (df[col].max() - df[col].min()) * 4
```

# 11. Tạo biến kết quả xét tuyển (KQXT)

```python
def kqxt(row):
    dh1, dh2, dh3 = row['DH1'], row['DH2'], row['DH3']
    if row['KT'] in ['A', 'A1']:
        return 1 if (dh1*2 + dh2 + dh3)/4 >= 5 else 0
    elif row['KT'] == 'B':
        return 1 if (dh1 + dh2*2 + dh3)/4 >= 5 else 0
    else:
        return 1 if (dh1 + dh2 + dh3)/3 >= 5 else 0

df['KQXT'] = df.apply(kqxt, axis=1)
```
