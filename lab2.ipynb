```python
def classify_t1(x):
    if x < 5:
        return 'k'
    elif 5 <= x < 7:
        return 'tb'
    elif 7 <= x < 8:
        return 'k'
    else:
        return 'g'

df['phanlopt1'] = df['T1'].apply(classify_t1)
```

```python
df['phanlopt1'].value_counts().plot(kind='bar', title='Tần số phân loại T1')
plt.show()
```
